```python
import openai

openai.api_key = "open_ai_key"
```

```python
def chat_with_gpt(prompt,history):
    responce = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": f"Sen bir hastane yönericisisin. Geçmiş konuşmalarımız: {history}, yeni soru: {prompt}"}])
    return responce.choices[0].message.content.strip()
```

```python
if __name__ == "__main__":
  history = []

  while True:

    user_input = input("Mesajınız nedir?")

    if user_input.lower() in ["exit",""]:
      print("Görüşmek tamamlandı")
      break

    history.append(user_input)
    responce = chat_with_gpt(user_input,history)
    print("Chatbot: ",responce)
```
